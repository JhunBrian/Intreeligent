```python
image_root = '../data/Cerrado_trees.v1i.coco/_separated'

data_dict = {'image_path': [], 'label':[]}
for img in os.listdir(image_root):
    img_path = os.path.join(image_root, img)
    label = re.search(r'_(?:png|jpg)_(.+?)_\d+\.png$', img_path)
    data_dict['image_path'].append(img_path)
    if label:
        data_dict['label'].append(label.group(1))

data_df = pd.DataFrame(data_dict)
data_df.head()
```

### **Autoencoder**

```python
ckpt_path = '../_Autoencoder/results/autoencoder_logs/version_3/checkpoints/autoencoder-epoch=29-val_mse=0.0000.ckpt'

model = LitAutoencoder.load_from_checkpoint(
    ckpt_path,
    architecture=ResNetAutoencoder(model_name="resnet34", pretrained=False)
)
```

```python
class ClusteringDataset(Dataset):
    def __init__(self, data_df, transforms=None):
        self.data_df = data_df
        self.transforms = transforms
        self.image_paths = self.data_df['image_path'].tolist()

    def __getitem__(self, idx):
        image = cv2.cvtColor(cv2.imread(self.image_paths[idx]), cv2.COLOR_BGR2RGB)
        if self.transforms:
            image = self.transforms(image)
        return image

    def __len__(self):
        return len(self.data_df['image_path'])
```

```python
trans = transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize(size=(512, 512)),
    MinMaxNormalize()
])

cd = ClusteringDataset(data_df, trans)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.eval()
model.to(device)


class LatentSpace:
    def __init__(self, model, device="cpu", return_recon=False):
        """
        model: your autoencoder model
        device: "cpu" or "cuda"
        return_recon: whether to return reconstruction along with latent
        """
        self.model = model.to(device)  # move model to device once
        self.device = device
        self.return_recon = return_recon
        
    def __call__(self, image):
        # move image to correct device
        image = image.to(self.device)
        return self.predict(image)

    def predict(self, image):
        self.model.eval()
        with torch.no_grad():
            # Add batch dimension if missing
            if image.dim() == 3:
                image = image.unsqueeze(0)

            image = image.to(self.device)  # ensure image is on same device as model

            # --- Encode ---
            z = self.model.model.encoder(image)

            if self.return_recon:
                # --- Reconstruct ---
                out = self.model.model.decoder(z)
                return z.cpu(), out.cpu()

            return z.cpu()
```

```python
ls = LatentSpace(model, 'cuda')
```

```python
latent_spaces = []
for image in tqdm_notebook(cd):
    latent_spaces.append(ls(image))
```

```python
import torch
import numpy as np
import umap
import hdbscan
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    adjusted_rand_score, normalized_mutual_info_score,
    homogeneity_score, completeness_score, v_measure_score,
    fowlkes_mallows_score
)
from scipy.spatial.distance import pdist, squareform
from tqdm import tqdm

class LatentCluster:
    def __init__(
        self,
        n_components=50,
        umap_n_neighbors=15,
        umap_min_dist=0.1,
        umap_metric='euclidean',
        min_cluster_size=5,
        min_samples=None,
        cluster_selection_epsilon=0.0,
        cluster_selection_method='eom',
        hdbscan_metric='euclidean',
        verbose=True
    ):
        # UMAP parameters
        self.n_components = n_components
        self.umap_n_neighbors = umap_n_neighbors
        self.umap_min_dist = umap_min_dist
        self.umap_metric = umap_metric

        # HDBSCAN parameters
        self.min_cluster_size = min_cluster_size
        self.min_samples = min_samples
        self.cluster_selection_epsilon = cluster_selection_epsilon
        self.cluster_selection_method = cluster_selection_method
        self.hdbscan_metric = hdbscan_metric

        self.verbose = verbose
        self.reducer = None
        self.clusterer = None
        self.X_reduced = None
        self.labels_ = None

    def _flatten_latents(self, latent_list):
        flattened = []
        for z in tqdm(latent_list, disable=not self.verbose):
            if z.dim() == 4:  # [1, C, H, W]
                flattened.append(z.view(-1).cpu().numpy())
            else:
                raise ValueError(f"Unexpected tensor shape: {z.shape}")
        return np.stack(flattened)

    def fit(self, latent_list):
        X = self._flatten_latents(latent_list)

        # Dimensionality reduction with UMAP
        self.reducer = umap.UMAP(
            n_components=self.n_components,
            n_neighbors=self.umap_n_neighbors,
            min_dist=self.umap_min_dist,
            metric=self.umap_metric
        )
        self.X_reduced = self.reducer.fit_transform(X)

        # Handle 'cosine' metric for HDBSCAN safely
        if self.hdbscan_metric == 'cosine':
            distance_matrix = squareform(pdist(self.X_reduced, metric='cosine'))
            self.clusterer = hdbscan.HDBSCAN(
                min_cluster_size=self.min_cluster_size,
                min_samples=self.min_samples,
                cluster_selection_epsilon=self.cluster_selection_epsilon,
                cluster_selection_method=self.cluster_selection_method,
                metric='precomputed'
            )
            self.labels_ = self.clusterer.fit_predict(distance_matrix)
        else:
            self.clusterer = hdbscan.HDBSCAN(
                min_cluster_size=self.min_cluster_size,
                min_samples=self.min_samples,
                cluster_selection_epsilon=self.cluster_selection_epsilon,
                cluster_selection_method=self.cluster_selection_method,
                metric=self.hdbscan_metric
            )
            self.labels_ = self.clusterer.fit_predict(self.X_reduced)

        return self

    def predict(self, latent_list):
        if self.reducer is None:
            raise ValueError("Model not fitted yet.")

        X = self._flatten_latents(latent_list)
        X_reduced = self.reducer.transform(X)

        # HDBSCAN cannot reliably predict new points
        labels = -1 * np.ones(X_reduced.shape[0], dtype=int)
        return labels

    def evaluate(self, y_true):
        if self.labels_ is None:
            raise ValueError("Fit the model first to get labels.")

        le = LabelEncoder()
        y_true_encoded = le.fit_transform(y_true)

        ari = adjusted_rand_score(y_true_encoded, self.labels_)
        nmi = normalized_mutual_info_score(y_true_encoded, self.labels_)
        homogeneity = homogeneity_score(y_true_encoded, self.labels_)
        completeness = completeness_score(y_true_encoded, self.labels_)
        v_measure = v_measure_score(y_true_encoded, self.labels_)
        fmi = fowlkes_mallows_score(y_true_encoded, self.labels_)

        metrics = {
            "ARI": ari,
            "NMI": nmi,
            "Homogeneity": homogeneity,
            "Completeness": completeness,
            "V-measure": v_measure,
            "FMI": fmi
        }
        return metrics


# --- Updated grid search helper ---
from itertools import product
import pandas as pd

def grid_search_latentcluster(latent_list, y_true, param_grid):
    keys, values = zip(*param_grid.items())
    results = []

    for combination in tqdm_notebook(product(*values), total=np.prod([len(v) for v in values])):
        params = dict(zip(keys, combination))

        lc = LatentCluster(
            n_components=params.get("n_components", 50),
            umap_n_neighbors=params.get("umap_n_neighbors", 15),
            umap_min_dist=params.get("umap_min_dist", 0.1),
            umap_metric=params.get("umap_metric", "euclidean"),
            min_cluster_size=params.get("min_cluster_size", 5),
            min_samples=params.get("min_samples", None),
            cluster_selection_epsilon=params.get("cluster_selection_epsilon", 0.0),
            cluster_selection_method=params.get("cluster_selection_method", "eom"),
            hdbscan_metric=params.get("hdbscan_metric", "euclidean"),
            verbose=False
        )

        lc.fit(latent_list)
        metrics = lc.evaluate(y_true)
        results.append({**params, **metrics})

    return pd.DataFrame(results)
```

```python
le = LabelEncoder() 
y_true = le.fit_transform(data_df['label'].tolist())
```

```python
param_grid = {
    # UMAP
    "n_components": [10, 50, 100],
    "umap_n_neighbors": [10, 15, 30],
    "umap_min_dist": [0.0, 0.1, 0.3],
    "umap_metric": ["euclidean", "cosine"],
    
    # HDBSCAN
    "min_cluster_size": [5, 10, 20],
    "min_samples": [None, 5],
    "cluster_selection_epsilon": [0.0, 0.1],
    "cluster_selection_method": ["eom", "leaf"],
    "hdbscan_metric": ["euclidean", "cosine"]
}

df_results = grid_search_latentcluster(latent_spaces, y_true, param_grid)
```

| Metric                                  | Measures                                                                           | Notes                                                    |
| --------------------------------------- | ---------------------------------------------------------------------------------- | -------------------------------------------------------- |
| **ARI** (Adjusted Rand Index)           | Pairwise agreement between predicted clusters and true labels, adjusted for chance | Sensitive to label permutations; 1 = perfect, 0 = random |
| **NMI** (Normalized Mutual Information) | Information shared between clusters and true labels                                | Normalized 0–1; good for imbalanced classes              |
| **Homogeneity**                         | Each cluster contains only members of a single class                               | 1 = perfect, 0 = mixed clusters                          |
| **Completeness**                        | All members of a class are assigned to the same cluster                            | 1 = perfect, 0 = split classes                           |
| **V-measure**                           | Harmonic mean of homogeneity and completeness                                      | 1 = perfect, balances the above two                      |
| **FMI** (Fowlkes–Mallows Index)         | Pairwise precision vs recall of clustering                                         | 1 = perfect, emphasizes pairwise agreement               |
